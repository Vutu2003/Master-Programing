### Problem 2.1 — Ước lượng phương sai khi biết trung bình

**Đề bài**

Với $N\ge 1$ và $x[0],\ldots,x[N-1]$ độc lập cùng phân phối (IID) $\mathcal N(0,\sigma^2)$, trong đó $\sigma^2>0$ chưa biết, xét
$$
\widehat{\sigma^2}=\frac1N\sum_{n=0}^{N-1}x[n]^2.
$$
Xác định unbiasedness, tính variance và xét giới hạn khi $N\to\infty$.

**Lời giải**

Moment bậc hai và bậc bốn của Gaussian tâm 0 lần lượt là $\mathbb E[x[n]^2]=\sigma^2$ và $\mathbb E[x[n]^4]=3\sigma^4$. Do đó
$$
\mathbb E[\widehat{\sigma^2}]=\frac1N\sum_{n=0}^{N-1}\sigma^2=\sigma^2,
\qquad
\operatorname{Var}(x[n]^2)=3\sigma^4-(\sigma^2)^2=2\sigma^4.
$$
Do các bình phương cũng độc lập,
$$
\operatorname{Var}(\widehat{\sigma^2})=\frac1{N^2}\sum_{n=0}^{N-1}2\sigma^4=\frac{2\sigma^4}{N}.
$$
Vậy $\widehat{\sigma^2}$ là unbiased estimator với mọi $\sigma^2>0$. Mean Squared Error (MSE) bằng variance $2\sigma^4/N$ và tiến về 0 với $\sigma^2$ cố định. Ước lượng hội tụ bình phương trung bình về $\sigma^2$, nên cũng hội tụ theo xác suất theo bất đẳng thức Markov áp dụng cho sai số bình phương.

**Ý nghĩa**

Biết trước trung bình bằng 0 cho phép dùng trực tiếp trung bình của các bình phương. Tăng $N$ làm giảm variance theo tốc độ $1/N$.

### Problem 2.2 — Ước lượng cận trên của phân phối đều

**Đề bài**

Cho $N\ge 1$ quan sát $x[0],\ldots,x[N-1]$ IID theo $\mathcal U[0,\theta]$, với $0<\theta<\infty$. Tìm một unbiased estimator của $\theta$.

**Lời giải**

Mỗi mẫu có Probability Density Function (PDF) bằng $1/\theta$ trên $[0,\theta]$ và bằng 0 ngoài khoảng đó. Vì vậy,
$$
\mathbb E_\theta[x[n]]=\frac1\theta\int_0^\theta u\,\mathrm{d}u=\frac{\theta}{2}.
$$
Với $\bar x=N^{-1}\sum_{n=0}^{N-1}x[n]$, tính tuyến tính của kỳ vọng cho
$$
\widehat\theta=2\bar x,\qquad
\mathbb E_\theta[\widehat\theta]=\frac2N\sum_{n=0}^{N-1}\frac{\theta}{2}=\theta
\qquad(\theta>0).
$$

**Ý nghĩa**

Trung bình của phân phối đều xác định cận trên qua $\theta=2\mathbb E_\theta[x[n]]$. Hệ số 2 chuyển trung bình mẫu thành ước lượng không chệch của cận trên.

### Problem 2.3 — PDF của trung bình mẫu Gaussian

**Đề bài**

Trong Ví dụ 2.1, $x[n]=A+w[n]$ với $n=0,\ldots,N-1$, $N\ge 1$, $A\in\mathbb R$ chưa biết và $w[n]$ là nhiễu trắng Gaussian (White Gaussian Noise, WGN) có trung bình 0, phương sai $\sigma^2>0$. Chứng minh $\widehat A=N^{-1}\sum_{n=0}^{N-1}x[n]$ có phân phối $\mathcal N(A,\sigma^2/N)$.

**Lời giải**

Vectơ nhiễu có phân phối Gaussian đồng thời. Tính trắng cho $\operatorname{Cov}(w[n],w[m])=0$ khi $n\ne m$, nên các thành phần độc lập. Tổ hợp tuyến tính $\widehat A=A+N^{-1}\sum_{n=0}^{N-1}w[n]$ là Gaussian, với
$$
\mathbb E[\widehat A]=A,\qquad
\operatorname{Var}(\widehat A)=\frac1{N^2}\sum_{n=0}^{N-1}\sigma^2
=\frac{\sigma^2}{N}.
$$
Suy ra $\widehat A\sim\mathcal N(A,\sigma^2/N)$ và
$$
f_{\widehat A}(a)=\sqrt{\frac{N}{2\pi\sigma^2}}\,
\exp\!\left(-\frac{N(a-A)^2}{2\sigma^2}\right),\qquad a\in\mathbb R.
$$

**Ý nghĩa**

Phân phối Gaussian này đúng với mọi cỡ mẫu, kể cả $N=1$. Độ lệch chuẩn của ước lượng bằng $\sigma/\sqrt N$.

### Problem 2.4 — Lấy trung bình các phép đo nhịp tim

**Đề bài**

Trong một giây, máy tính ghi 10 phép đo nhịp tim $h$, mỗi phép đo cách nhau $100\,\mathrm{ms}$. Các phép đo $\widehat h_1,\ldots,\widehat h_{10}$ đôi một không tương quan, với $\mathbb E[\widehat h_i]=\alpha h$ và $\operatorname{Var}(\widehat h_i)=1$. Với $\alpha=1$ và $\alpha=1/2$, xét việc lấy trung bình $\widehat h=10^{-1}\sum_{i=1}^{10}\widehat h_i$ có cải thiện ước lượng $h$ so với một phép đo hay không.

**Lời giải**

Lấy Mean Squared Error (MSE) đối với $h$ làm tiêu chí so sánh. Cả $\widehat h_i$ và $\widehat h$ đều có kỳ vọng $\alpha h$, nên có cùng bias $(\alpha-1)h$. Các hiệp phương sai chéo bằng 0, do đó
$$
\operatorname{Var}(\widehat h)=\frac{1}{10^2}\sum_{i=1}^{10}\operatorname{Var}(\widehat h_i)
=\frac1{10}.
$$
Dùng $\operatorname{MSE}(T)=\mathbb E[(T-h)^2]=\operatorname{Var}(T)+(\mathbb E[T]-h)^2$ cho hai trường hợp.

| $\alpha$ | Bias của mỗi ước lượng | $\operatorname{MSE}(\widehat h_i)$ | $\operatorname{MSE}(\widehat h)$ |
| --- | ---: | ---: | ---: |
| $1$ | $0$ | $1$ | $1/10$ |
| $1/2$ | $-h/2$ | $1+h^2/4$ | $1/10+h^2/4$ |

Lấy trung bình giảm MSE đúng $9/10$ trong cả hai trường hợp.

**Ý nghĩa**

Với $\alpha=1$, lấy trung bình giảm variance của một unbiased estimator. Với $\alpha=1/2$, variance cũng giảm và bias $-h/2$ vẫn còn.

### Problem 2.5 — PDF của ước lượng phương sai từ hai mẫu

**Đề bài**

Cho $x[0],x[1]$ độc lập theo $\mathcal N(0,\sigma^2)$, $\sigma^2>0$. Tìm PDF của unbiased estimator $\widehat{\sigma^2}=(x[0]^2+x[1]^2)/2$ và xét tính đối xứng quanh $\sigma^2$.

**Lời giải**

Hai biến $x[0]/\sigma$ và $x[1]/\sigma$ độc lập theo $\mathcal N(0,1)$, nên
$$
Y=\frac{x[0]^2+x[1]^2}{\sigma^2}\sim\chi^2_2,\qquad
f_Y(y)=\tfrac12\exp(-y/2)\quad (y\ge 0).
$$
Đặt $T=\widehat{\sigma^2}=(\sigma^2/2)Y$. Phép đổi biến $y=2t/\sigma^2$ có Jacobian $2/\sigma^2$, nên
$$
f_T(t)=
\begin{cases}
\dfrac{1}{\sigma^2}\exp(-t/\sigma^2), & t\ge 0,\\
0, & t<0.
\end{cases}
$$
Đây là phân phối mũ với scale $\sigma^2$, support $[0,\infty)$ và kỳ vọng $\sigma^2$. Vì
$$
\Pr(T<\sigma^2)=1-e^{-1}>\tfrac12,\qquad
\Pr(T>\sigma^2)=e^{-1},
$$
PDF không đối xứng quanh $\sigma^2$.

**Kiểm chứng số**

Với $\sigma^2=2$, mô phỏng $R=60\,000$ cặp Gaussian độc lập. So sánh histogram với PDF mũ chính xác, đồng thời kiểm tra $\mathbb E[T]$ và $p=\Pr(T<\sigma^2)$. Sai số chuẩn Monte Carlo của trung bình và tỉ lệ lần lượt là $\sigma^2/\sqrt R$ và $\sqrt{p(1-p)/R}$.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import expon

rng = np.random.default_rng(205)
noise_variance = 2.0
num_trials = 60_000

samples = rng.normal(scale=np.sqrt(noise_variance), size=(num_trials, 2))
variance_estimates = np.mean(samples**2, axis=1)
sampling_distribution = expon(scale=noise_variance)

observed_mean = variance_estimates.mean()
observed_fraction = (variance_estimates < noise_variance).mean()
expected_fraction = sampling_distribution.cdf(noise_variance)
mean_mcse = noise_variance / np.sqrt(num_trials)
fraction_mcse = np.sqrt(
    expected_fraction * (1 - expected_fraction) / num_trials
)

print(f"{'Đại lượng':<24}{'Mô phỏng':>12}{'Lý thuyết':>12}{'SE MC':>12}")
print(
    f"{'Trung bình':<24}{observed_mean:>12.5f}"
    f"{noise_variance:>12.5f}{mean_mcse:>12.5f}"
)
print(
    f"{'Tỉ lệ dưới σ²':<24}{observed_fraction:>12.5f}"
    f"{expected_fraction:>12.5f}{fraction_mcse:>12.5f}"
)
assert abs(observed_mean - noise_variance) <= 4 * mean_mcse
assert abs(observed_fraction - expected_fraction) <= 4 * fraction_mcse

grid = np.linspace(0.0, variance_estimates.max(), 400)
_, ax = plt.subplots(figsize=(7, 3.8), layout="constrained")
ax.hist(
    variance_estimates,
    bins=75,
    density=True,
    alpha=0.45,
    label="Mô phỏng",
)
ax.plot(grid, sampling_distribution.pdf(grid), label="PDF mũ chính xác")
ax.axvline(noise_variance, color="0.3", linestyle="--", label=r"$\sigma^2$")
ax.set(
    xlabel=r"$\widehat{\sigma^2}$",
    ylabel="Mật độ",
    xlim=(0, 6 * noise_variance),
)
ax.legend()
plt.show()
```

**Ý nghĩa**

Histogram có đuôi phải như PDF mũ. Với seed đã chọn, trung bình và tỉ lệ mô phỏng đều cách giá trị lý thuyết dưới bốn sai số chuẩn Monte Carlo. Tính không chệch xác định kỳ vọng, còn ở đây khoảng $63{,}2\%$ số ước lượng nằm dưới giá trị thật.

### Problem 2.6 — Trọng số tuyến tính có variance nhỏ nhất

**Đề bài**

Trong mô hình Ví dụ 2.1, $x[n]=A+w[n]$ với $n=0,\ldots,N-1$, $N\ge 1$, $A\in\mathbb R$ và $w[n]$ IID theo $\mathcal N(0,\sigma^2)$, $\sigma^2>0$. Tìm các trọng số thực cố định $a_n$ sao cho $\widehat A=\sum_{n=0}^{N-1}a_nx[n]$ là unbiased estimator có variance nhỏ nhất trong lớp này.

**Lời giải**

Ta có $\mathbb E[\widehat A]=A\sum_n a_n$ và, do các mẫu độc lập, $\operatorname{Var}(\widehat A)=\sigma^2\sum_n a_n^2$. Unbiasedness với mọi $A\in\mathbb R$ tương đương $\sum_n a_n=1$. Bài toán là
$$
\min\left\{\sigma^2\sum_{n=0}^{N-1}a_n^2 \ \middle| \ 
a\in\mathbb R^N,\ \sum_{n=0}^{N-1}a_n=1\right\}.
$$
Đặt $\mathcal L(a,\lambda)=\sigma^2\sum_n a_n^2-\lambda(\sum_n a_n-1)$. Điều kiện dừng theo từng $a_n$ cho $2\sigma^2a_n-\lambda=0$, nên mọi trọng số bằng nhau. Ràng buộc tổng bằng 1 cho
$$
a_n=\frac1N\quad(n=0,\ldots,N-1),\qquad
\operatorname{Var}_{\min}=\frac{\sigma^2}{N}.
$$
Vì hàm mục tiêu lồi nghiêm ngặt ($\sigma^2>0$) và ràng buộc là affine, điểm dừng thỏa ràng buộc là nghiệm tối ưu toàn cục duy nhất.

**Ý nghĩa**

Khi các phép đo không tương quan và cùng phương sai, trung bình mẫu là ước lượng có variance nhỏ nhất trong lớp ước lượng tuyến tính không chệch.

### Problem 2.7 — Xác suất sai số lớn của ước lượng Gaussian

**Đề bài**

Hai unbiased estimator Gaussian $\widehat\theta$ và $\check\theta$ của $\theta$ có $\operatorname{Var}(\widehat\theta)<\operatorname{Var}(\check\theta)$. Chứng minh rằng với mọi $\varepsilon>0$,
$$
\Pr_\theta\!\left(|\widehat\theta-\theta|>\varepsilon\right)
<
\Pr_\theta\!\left(|\check\theta-\theta|>\varepsilon\right).
$$

**Lời giải**

Cố định $\theta$ và gọi $s_1,s_2$ là hai độ lệch chuẩn, $s_1<s_2$. Do unbiasedness, mỗi sai số Gaussian có kỳ vọng 0. Với sai số $U\sim\mathcal N(0,s^2)$ và $s>0$, chuẩn hóa cho
$$
\Pr\!\left(|U|>\varepsilon\right)
=2\left[1-\Phi\!\left(\frac{\varepsilon}{s}\right)\right],
$$
trong đó $\Phi$ là hàm phân phối tích lũy (CDF) của $\mathcal N(0,1)$. Nếu $s_1>0$ thì $\varepsilon/s_1>\varepsilon/s_2$ và $\Phi$ tăng nghiêm ngặt, nên xác suất đuôi của $\widehat\theta$ nhỏ hơn. Nếu cho phép Gaussian suy biến với $s_1=0$, xác suất bên trái bằng 0 còn bên phải dương vì $s_2>0$.

**Ý nghĩa**

Trong lớp Gaussian cùng tâm tại giá trị thật, variance nhỏ hơn cho xác suất vượt mọi ngưỡng sai số tuyệt đối nhỏ hơn. So sánh này chỉ dùng phân phối biên của hai ước lượng.

### Problem 2.8 — Tính nhất quán của trung bình mẫu

**Đề bài**

Cho $x[n]=A+w[n]$ như trong Ví dụ 2.1, với $A\in\mathbb R$ cố định và $w[n]$ IID theo $\mathcal N(0,\sigma^2)$, $\sigma^2>0$. Dùng kết quả Problem 2.3 để chứng minh rằng $\Pr_A(|\widehat A-A|>\varepsilon)\to 0$ khi $N\to\infty$ với mọi $\varepsilon>0$, trong đó $\widehat A=N^{-1}\sum_{n=0}^{N-1}x[n]$. Xét thêm $\check A=(2N)^{-1}\sum_{n=0}^{N-1}x[n]$.

**Lời giải**

Theo Problem 2.3, $\widehat A-A\sim\mathcal N(0,\sigma^2/N)$. Gọi $\Phi$ là CDF của Gaussian chuẩn và $\bar\Phi(z)=1-\Phi(z)$ là xác suất đuôi phải. Với mọi $\varepsilon>0$,
$$
\Pr_A\!\left(|\widehat A-A|>\varepsilon\right)
=2\bar\Phi\!\left(\frac{\varepsilon\sqrt N}{\sigma}\right)
\longrightarrow 0.
$$
Vậy $\widehat A\xrightarrow{\mathsf P} A$, tức hội tụ theo xác suất (convergence in probability) về $A$.

Với $\check A=\widehat A/2$, ta có $\check A\sim\mathcal N(A/2,\sigma^2/(4N))$. Đặt $d=A/2$ và $s_N=\sigma/(2\sqrt N)$. Sai số $\check A-A$ có phân phối $\mathcal N(-d,s_N^2)$, nên
$$
\Pr_A\!\left(|\check A-A|>\varepsilon\right)
=\Phi\!\left(\frac{d-\varepsilon}{s_N}\right)
 +\bar\Phi\!\left(\frac{d+\varepsilon}{s_N}\right).
$$
Vì $s_N\to0$, giới hạn được xác định bởi dấu của $d-\varepsilon$ và $d+\varepsilon$, với $\Phi(0)=\bar\Phi(0)=1/2$.
$$
\lim_{N\to\infty}\Pr_A\!\left(|\check A-A|>\varepsilon\right)
=\begin{cases}
1, & 0<\varepsilon<|A|/2,\\
1/2, & \varepsilon=|A|/2>0,\\
0, & \varepsilon>|A|/2.
\end{cases}
$$
Vì $\Pr_A(|\check A-A/2|>\varepsilon)=\Pr_A(|\widehat A-A|>2\varepsilon)\to0$ nên $\check A\xrightarrow{\mathsf P} A/2$ và chỉ hội tụ về giá trị cần ước lượng tại $A=0$.

**Kiểm chứng số**

Với $A=2$ và $\sigma=1$, tính trực tiếp xác suất đuôi Gaussian hữu hạn mẫu. Các ngưỡng $\varepsilon=0{,}5,\ 1,\ 1{,}5$ của $\check A$ nằm lần lượt dưới, tại và trên $|A|/2=1$. Đường của $\widehat A$ ở ngưỡng $0{,}5$ cho phép so sánh hai ước lượng tại cùng mức sai số.

```python
from scipy.stats import norm

amplitude = 2.0
noise_sd = 1.0
sample_sizes = np.arange(1, 257)
mean_sd = noise_sd / np.sqrt(sample_sizes)
half_mean_sd = mean_sd / 2
offset = amplitude / 2
thresholds = (0.5, 1.0, 1.5)

_, ax = plt.subplots(figsize=(7, 3.8), layout="constrained")
for epsilon in thresholds:
    tail_probability = (
        norm.cdf((offset - epsilon) / half_mean_sd)
        + norm.sf((offset + epsilon) / half_mean_sd)
    )
    ax.semilogx(
        sample_sizes,
        tail_probability,
        label=rf"$\check A$, $\varepsilon={epsilon:g}$",
    )

reference_threshold = thresholds[0]
ax.semilogx(
    sample_sizes,
    2 * norm.sf(reference_threshold / mean_sd),
    color="0.25",
    linestyle="--",
    label=rf"$\widehat A$, $\varepsilon={reference_threshold:g}$",
)
ax.set(
    xlabel="Cỡ mẫu $N$",
    ylabel="Xác suất vượt ngưỡng",
    ylim=(-0.025, 1.025),
)
ax.legend(loc="center right", bbox_to_anchor=(1, 0.76))
plt.show()
```

**Ý nghĩa**

Ba đường của $\check A$ cho thấy các giới hạn $1$, $1/2$ và $0$ dù variance đều giảm về 0. Tính nhất quán (consistency) đòi hỏi sự tập trung quanh giá trị cần ước lượng $A$, trong khi tâm của $\check A$ là $A/2$.

### Problem 2.9 — Bias của phép bình phương trung bình mẫu

**Đề bài**

Trong Ví dụ 2.1, $x[n]=A+w[n]$ với $n=0,\ldots,N-1$, $N\ge 1$, $A\in\mathbb R$ chưa biết và $w[n]$ IID theo $\mathcal N(0,\sigma^2)$, $\sigma^2>0$. Xét $\widehat\theta=\widehat A^2$ để ước lượng $\theta=A^2$, trong đó $\widehat A=N^{-1}\sum_{n=0}^{N-1}x[n]$. Xác định bias và hành vi khi $N\to\infty$.

**Lời giải**

Đặt $Z_N=\widehat A-A$ và $v_N=\sigma^2/N$. Khi đó $Z_N\sim\mathcal N(0,v_N)$, nên
$$
\mathbb E[\widehat\theta]=\mathbb E[A^2+2AZ_N+Z_N^2]=A^2+v_N.
$$
Vậy bias là $\sigma^2/N>0$ với mọi $N$ hữu hạn và tiến về 0 khi $N\to\infty$. Từ $\mathbb E[Z_N^3]=0$ và $\mathbb E[Z_N^4]=3v_N^2$, ta có
$$
\begin{aligned}
\operatorname{MSE}(\widehat\theta)
&=\mathbb E[(2AZ_N+Z_N^2)^2]\\
&=4A^2v_N+4A\mathbb E[Z_N^3]+\mathbb E[Z_N^4]
=4A^2v_N+3v_N^2\longrightarrow 0.
\end{aligned}
$$
Do đó $\widehat\theta$ hội tụ bình phương trung bình và hội tụ theo xác suất về $A^2$.

**Ý nghĩa**

Phép bình phương đưa variance của $\widehat A$ vào kỳ vọng của $\widehat\theta$. Ước lượng $A^2$ này có bias ở cỡ mẫu hữu hạn và vẫn nhất quán.

### Problem 2.10 — Unbiasedness của ước lượng vectơ

**Đề bài**

Trong Ví dụ 2.1, giờ cả $A\in\mathbb R$ và $\sigma^2>0$ đều chưa biết. Quan sát $x[0],\ldots,x[N-1]$ IID theo $\mathcal N(A,\sigma^2)$ với $N\ge 2$. Xét tính không chệch của vectơ ước lượng
$$
\widehat{\boldsymbol\theta}
=\begin{bmatrix}\widehat A\\\widehat{\sigma^2}\end{bmatrix},
\qquad
\widehat A=\frac1N\sum_{n=0}^{N-1}x[n],\qquad
\widehat{\sigma^2}=\frac1{N-1}\sum_{n=0}^{N-1}(x[n]-\widehat A)^2.
$$
Bản in trang 24 đặt dấu mũ trên ký hiệu vectơ tham số thật. Ở đây ký hiệu $\boldsymbol\theta=(A,\sigma^2)^{\mathsf T}$ chỉ tham số, còn $\widehat{\boldsymbol\theta}$ chỉ ước lượng.

**Lời giải**

Tính tuyến tính của kỳ vọng và tính độc lập cho $\mathbb E[\widehat A]=A$ và $\operatorname{Var}(\widehat A)=\sigma^2/N$. Khai triển $x[n]-A=(x[n]-\widehat A)+(\widehat A-A)$ và dùng $\sum_{n=0}^{N-1}(x[n]-\widehat A)=0$ để có
$$
\sum_{n=0}^{N-1}(x[n]-\widehat A)^2
=\sum_{n=0}^{N-1}(x[n]-A)^2-N(\widehat A-A)^2.
$$
Lấy kỳ vọng cho $N\sigma^2-N(\sigma^2/N)=(N-1)\sigma^2$ ở vế phải. Do đó $\mathbb E[\widehat{\sigma^2}]=\sigma^2$ và
$$
\mathbb E_{\boldsymbol\theta}[\widehat{\boldsymbol\theta}]
=(A,\sigma^2)^{\mathsf T}=\boldsymbol\theta
\qquad ((A,\sigma^2)\in\mathbb R\times(0,\infty)).
$$

**Ý nghĩa**

Unbiasedness của vectơ được kiểm tra theo từng thành phần. Thay $A$ bằng $\widehat A$ làm giảm kỳ vọng tổng bình phương đi $\sigma^2$, do đó mẫu số hiệu chỉnh là $N-1$.

### Problem 2.11 — Không tồn tại unbiased estimator từ một mẫu

**Đề bài**

Cho một quan sát $x[0]\sim\mathcal U[0,1/\theta]$ với $\theta>0$. Chứng minh rằng unbiasedness của $\widehat\theta=g(x[0])$ đòi hỏi $\int_0^{1/\theta}g(u)\,\mathrm du=1$. Sau đó chứng minh không có hàm $g$ thỏa điều kiện này với mọi $\theta>0$.

**Lời giải**

PDF có giá trị $\theta$ trên $[0,1/\theta]$ và bằng 0 ở ngoài. Với một hàm đo được $g$ có kỳ vọng hữu hạn, unbiasedness tương đương
$$
\mathbb E_\theta[g(x[0])]
=\theta\int_0^{1/\theta}g(u)\,\mathrm du=\theta
\quad\Longleftrightarrow\quad
\int_0^{1/\theta}g(u)\,\mathrm du=1.
$$
Đặt $t=1/\theta$. Vì $\theta$ chạy khắp $(0,\infty)$, điều kiện trên buộc $\int_0^t g(u)\,\mathrm du=1$ với mọi $t>0$.

Đặc biệt, kỳ vọng hữu hạn tại $\theta=1$ đòi hỏi $\int_0^1|g(u)|\,\mathrm du<\infty$. Khi $t\downarrow0$, hàm $g(u)\mathbf{1}_{(0,t]}(u)$ hội tụ về 0 hầu khắp nơi trên $(0,1)$ và bị chặn theo trị tuyệt đối bởi hàm khả tích $|g(u)|$. Theo định lý hội tụ bị chặn (dominated convergence),
$$
\lim_{t\downarrow0}\int_0^t g(u)\,\mathrm du=0.
$$
Điều này mâu thuẫn với việc mọi tích phân đều bằng 1, nên không tồn tại unbiased estimator dạng $g(x[0])$.

**Ý nghĩa**

Khi $\theta$ tăng, miền quan sát co về 0. Unbiasedness ở một giá trị tham số đã buộc $g$ khả tích gần 0, khiến tích phân trên miền đang co phải tiến về 0.

## Final Summary

- Unbiasedness yêu cầu kỳ vọng đúng tại mọi giá trị tham số. Một unbiased estimator có thể có PDF lệch, và phép biến đổi phi tuyến có thể tạo bias.
- Tách $\operatorname{MSE}=\operatorname{Var}+\operatorname{Bias}^2$ cho thấy lấy trung bình giảm phần dao động của các phép đo không tương quan. Consistency còn phụ thuộc vào nơi phân phối tập trung khi $N$ tăng.
- Tối ưu hóa variance cần nêu rõ lớp ước lượng. Trong mô hình nhiễu trắng cùng phương sai, trọng số đều là nghiệm duy nhất trong lớp tuyến tính không chệch.
- Điều kiện unbiasedness có thể vô nghiệm. Với một quan sát đều trên $[0,1/\theta]$, tính khả tích gần 0 loại trừ khả năng ước lượng không chệch $\theta$.
